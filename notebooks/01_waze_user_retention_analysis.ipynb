# Waze User Churn Exploratory Data Analysis

## Objective
The goal of this analysis is to explore user driving behavior data from Waze and identify patterns associated with user churn.

## Key Questions
1. Are there missing values and do they follow a pattern?
2. How do churned users differ from retained users?
3. Does device type influence churn?
4. What behavioral metrics best explain churn?

### **01. Imports and data loading**

```python
# Import packages for data manipulation
import pandas as pd
import numpy as np
```

```python
# Load dataset into dataframe
df = pd.read_csv('waze_dataset.csv')
```

## Dataset
The dataset contains **14,999 users** and **13 behavioral variables**, including:
- sessions
- drives
- kilometers driven
- driving duration
- activity days
- device type

### **03. Null values and summary statistics**


```python
# rows with null labels
null_rows = df[df['label'].isna()]

# summary stats
null_rows.describe()
```

```python
# rows without null labels
non_null_rows = df[df['label'].notna()]

# summary stats
non_null_rows.describe()
```

The summary statistics of the rows with missing labels are very similar to those of the rows without missing values. Metrics such as sessions, drives, kilometers driven, and driving duration show comparable averages between the two groups. This suggests that the missing label values are likely random and not associated with a specific pattern in the dataset.

| Metric                  | Null label mean | Non-null mean |
| ----------------------- | --------------- | ------------- |
| sessions                | ~80.84          | ~80.62        |
| drives                  | ~67.80          | ~67.26        |
| driven_km_drives        | ~3936 km        | ~4044 km      |
| duration_minutes_drives | ~1795           | ~1864         |

### **04. Null values - device counts**

```python
# Get count of null values by device
null_rows['device'].value_counts()
```

Among the rows with missing values in the label column, 447 users were using an iPhone and 253 users were using Android devices.

```python
# Calculate % of iPhone nulls and Android nulls
null_rows['device'].value_counts(normalize=True)
```

```python
# Calculate % of iPhone users and Android users in full dataset
df['device'].value_counts(normalize=True)
```

The percentage of missing values by each device is consistent with their representation in the data overall.

There is nothing to suggest a non-random cause of the missing data.

Counts and percentages of users who churned vs. those who were retained.

```python
# Calculate counts of churned vs. retained
df['label'].value_counts()

```

```python
df['label'].value_counts(normalize=True)
```

This dataset contains 82% retained users and 18% churned users.

"Median" comparison of each variable for churned and retained users.

```python
# Calculate median values of all columns for churned and retained users
df.groupby('label').median(numeric_only=True)
```

This offers an interesting snapshot of the two groups, churned vs. retained:

Users who churned averaged ~3 more drives in the last month than retained users, but retained users used the app on over twice as many days as churned users in the same time period.

The median churned user drove ~200 more kilometers and 2.5 more hours during the last month than the median retained user.

It seems that churned users had more drives in fewer days, and their trips were farther and longer in duration. Perhaps this is suggestive of a user profile.

"Median" kilometers per drive in the last month for both retained and churned users.

```python
# Add a column to df called `km_per_drive`
df['km_per_drive'] = df['driven_km_drives'] / df['drives']

# Group by `label`, calculate the median, and isolate for km per drive
df.groupby('label')['km_per_drive'].median()
```

The median retained user drove about one more kilometer per drive than the median churned user.

How many kilometers per driving day was this?

```python
# Add a column to df called `km_per_driving_day`
df['km_per_driving_day'] = df['driven_km_drives'] / df['driving_days']

# Group by `label`, calculate the median, and isolate for km per driving day
df.groupby('label')['km_per_driving_day'].median()
```

"Median" number of drives per driving day for each group.

```python
# Add a column to df called `drives_per_driving_day`
df['drives_per_driving_day'] = df['drives'] / df['driving_days']

# Group by `label`, calculate the median, and isolate for drives per driving day
df.groupby('label')['drives_per_driving_day'].median()
```

The median user who churned drove 698 kilometers each day they drove last month, which is almost ~240% the per-drive-day distance of retained users. The median churned user had a similarly disproporionate number of drives per drive day compared to retained users.

It is clear from these figures that, regardless of whether a user churned or not, the users represented in this data are serious drivers! It would probably be safe to assume that this data does not represent typical drivers at large. Perhaps the data&mdash;and in particular the sample of churned users&mdash;contains a high proportion of long-haul truckers.

In consideration of how much these users drive, it would be worthwhile to recommend to Waze that they gather more data on these super-drivers. It's possible that the reason for their driving so much is also the reason why the Waze app does not meet their specific set of needs, which may differ from the needs of a more typical driver, such as a commuter.

Check for imbalance in how many users churned by device type?

```python
# For each label, calculate the number of Android users and iPhone users
df.groupby(['label', 'device']).size()
```

```python
# For each label, calculate the percentage of Android users and iPhone users
df.groupby('label')['device'].value_counts(normalize=True)
```

The ratio of iPhone users and Android users is consistent between the churned group and the retained group, and those ratios are both consistent with the ratio found in the overall dataset.

The distribution of device types is very similar between churned and retained users. Approximately 65% of users in both groups use iPhones and about 35% use Android devices. Because these proportions are nearly identical, device type does not appear to significantly influence whether a user churns or remains active.

### **Key Findings**

1. The dataset contains missing values in the label column. Out of 14,999 rows, 700 rows were missing churn labels. When comparing users with missing labels to those with complete data, the summary statistics were very similar. Additionally, the distribution of device types among rows with missing values closely matched the distribution in the overall dataset. This suggests that the missing data is likely random rather than systematic.

2. The median is less sensitive to extreme values and outliers. In this dataset, some users drove extremely large distances, which could distort the mean. Using the median provides a more accurate representation of the typical user behavior.

3. The analysis suggests that many users in the dataset may be heavy or long-distance drivers, possibly professional drivers such as truck or delivery drivers. It would be useful to investigate whether these users have different navigation needs compared to typical commuters. Additional information about user occupation, trip purpose, or driving context could help better understand why some users churn.

4. Approximately 35% of users were Android users and about 65% were iPhone users in the dataset.

5. Churned users tended to:

    Drive longer distances per day

    Have more drives per driving day

    Spend more total time driving

   However, retained users used the app on more days overall, suggesting more consistent engagement with the application.

6. The proportion of Android and iPhone users was nearly identical between churned and retained users. This indicates that device type does not appear to significantly influence churn.
