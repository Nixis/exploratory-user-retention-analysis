```python
df.columns
```

```python
# 1. Create `map_dictionary`

map_dictionary = {'iPhone': 1, 'Android': 2}

# 2. Create new `device_type` column

df['device_type'] = df['device']

# 3. Map the new column to the dictionary

df['device_type'] = df['device_type'].map(map_dictionary)
```

```python
df.groupby('device')['drives'].mean()
```

Based on the averages shown, it appears that drivers who use an iPhone device to interact with the application have a higher number of drives on average. However, this difference might arise from random sampling, rather than being a true difference in the number of drives. To assess whether the difference is statistically significant, conducting a hypothesis test seems appropriate.


### **03. Hypothesis testing**

Null hypothesis (H₀): There is no difference in the average number of drives between iPhone users and Android users.

Alternative hypothesis (Hₐ): There is a difference in the average number of drives between iPhone users and Android users.

Significance level = 5%

```python
# 1. Isolate the `drives` column for iPhone users.
iphone = df[df['device'] == 'iPhone']['drives']

# 2. Isolate the `drives` column for Android users.
android = df[df['device'] == 'Android']['drives']

# 3. Perform the t-test
stats.ttest_ind(a=iphone, b=android, equal_var=False)
```

Because the p-value is greater than the significance level of 0.05, the test fail to reject the null hypothesis.

This means there is not enough statistical evidence to conclude that the average number of drives is different between iPhone and Android users.

### **04. Key findings **

The hypothesis test showed that there is no statistically significant difference in the average number of drives between iPhone and Android users. Although iPhone users had a slightly higher average number of drives, the difference is small and could be due to random variation in the sample.

This suggests that device type alone may not be a strong factor influencing driving activity. Waze stakeholders may gain more useful insights by analyzing other variables such as activity days, sessions, driving days, or user retention patterns.
